# <font color='red'>Define parameters</font>

**Parameters**

- <font color='blue'>dataset type</font> – **attenuation** or **transmission** where attenuation = -log(transmission). Defaults to ‘attenuation’.

- <font color='blue'>subspace dimension</font> – Desired dimension of the subspace/ If None, the dimension is either set from the provided subspace basis matrix or estimated automatically from the data. Defaults to None.

- <font color='blue'>safety factor</font> – Multiplicative factor ≥ 1 used to scale the initial estimate of subspace dimension and ensure safer final choice. Defaults to 2.

- <font color='blue'>beta loss</font> – Beta divergence minimized in NMF. Can be **frobenius** or **kullback-leibler**. Defaults to ‘frobenius’.

- <font color='blue'>max iter</font> – Maximum iterations for the NMF solver. Defaults to 300.
