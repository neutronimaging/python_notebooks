```python
from skimage.io import imread
import os
import numpy as np
from PIL import Image
import dxchange

import matplotlib.pyplot as plt
```

```python
tif_image = "tests/data/images/tiff/image001.tif"
os.path.exists(tif_image)
```

```python
# using imread to read the tif image
image_with_imread = imread(tif_image).astype(np.float32)
```

```python
# using PIL
image_with_pil = np.array(Image.open(tif_image)).astype(np.float32)
```

```python
# using dxchange
image_with_dxchange = dxchange.read_tiff(tif_image).astype(np.float32)
```

```python
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
axs[0].imshow(image_with_imread, cmap='gray')
axs[0].set_title('Image read with imread')
axs[1].imshow(image_with_pil, cmap='gray')
axs[1].set_title('Image read with PIL')
axs[2].imshow(image_with_dxchange, cmap='gray')
axs[2].set_title('Image read with dxchange')
```
